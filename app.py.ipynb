```python
# ============================================================
# PART 6H — SCENARIO SIMULATOR
# ============================================================

from pathlib import Path

PROJECT_DIR = Path.cwd()
APP_FILE = PROJECT_DIR / "app.py"

app_code = APP_FILE.read_text(encoding="utf-8")

marker = 'elif page == "Scenario Simulator":'

if marker not in app_code:

    print("ERROR: Scenario Simulator section not found.")
    print()
    print("Available pages:")

    for line in app_code.splitlines():
        if "elif page ==" in line:
            print(line)

else:

    start = app_code.index(marker)

    remaining = app_code[start + len(marker):]

    next_marker = remaining.find("\nelif page ==")

    if next_marker == -1:
        end = len(app_code)
    else:
        end = start + len(marker) + next_marker

    # ========================================================
    # SCENARIO SIMULATOR PAGE
    # ========================================================

    scenario_page = r'''
elif page == "Scenario Simulator":

    st.title("🌍 Scenario Simulator")

    st.caption(
        "Assumption-driven emissions projections from 2024 to 2050."
    )

    # ========================================================
    # LOAD SCENARIO DATA
    # ========================================================

    @st.cache_data
    def load_global_scenarios():

        path = BASE_DIR / "outputs" / "global_scenario_projections.csv"

        if not path.exists():
            return pd.DataFrame()

        return pd.read_csv(path)


    @st.cache_data
    def load_scenario_assumptions():

        path = BASE_DIR / "outputs" / "scenario_assumptions.csv"

        if not path.exists():
            return pd.DataFrame()

        return pd.read_csv(path)


    @st.cache_data
    def load_key_projection_years():

        path = BASE_DIR / "outputs" / "key_projection_years.csv"

        if not path.exists():
            return pd.DataFrame()

        return pd.read_csv(path)


    @st.cache_data
    def load_country_scenarios():

        path = BASE_DIR / "outputs" / "country_2050_scenarios.csv"

        if not path.exists():
            return pd.DataFrame()

        return pd.read_csv(path)


    @st.cache_data
    def load_sector_scenarios():

        path = BASE_DIR / "outputs" / "sector_2050_scenarios.csv"

        if not path.exists():
            return pd.DataFrame()

        return pd.read_csv(path)


    @st.cache_data
    def load_country_sector_scenarios():

        path = (
            BASE_DIR
            / "outputs"
            / "country_sector_scenario_projections.csv"
        )

        if not path.exists():
            return pd.DataFrame()

        return pd.read_csv(path)


    global_scenarios = load_global_scenarios()
    assumptions = load_scenario_assumptions()
    key_projection_years = load_key_projection_years()
    country_scenarios = load_country_scenarios()
    sector_scenarios = load_sector_scenarios()
    country_sector_scenarios = load_country_sector_scenarios()


    # ========================================================
    # VALIDATION
    # ========================================================

    if global_scenarios.empty:

        st.error(
            "Global scenario data is missing. "
            "Expected outputs/global_scenario_projections.csv"
        )

        st.stop()


    # ========================================================
    # SCENARIO DEFINITIONS
    # ========================================================

    st.subheader("1. Scenario Assumptions")

    scenario_order = [
        "Baseline",
        "Moderate Reduction",
        "Accelerated Reduction"
    ]

    if not assumptions.empty:

        assumptions_display = assumptions.copy()

        if "Scenario" in assumptions_display.columns:

            assumptions_display["Scenario"] = pd.Categorical(
                assumptions_display["Scenario"],
                categories=scenario_order,
                ordered=True
            )

            assumptions_display = (
                assumptions_display
                .sort_values("Scenario")
                .reset_index(drop=True)
            )

        st.dataframe(
            assumptions_display,
            use_container_width=True,
            hide_index=True
        )

    else:

        st.warning(
            "Scenario assumptions file was not found."
        )


    st.markdown("""
    **Interpretation**

    These are structured scenarios rather than predictions.

    The scenarios modify the historical growth trajectory using
    explicit annual adjustments.

    They should therefore be interpreted as **what-if projections**
    rather than guaranteed future outcomes.
    """)


    # ========================================================
    # GLOBAL SCENARIO SELECTOR
    # ========================================================

    st.subheader("2. Global Scenario")

    available_scenarios = [
        s for s in scenario_order
        if s in global_scenarios["Scenario"].unique()
    ]

    selected_scenario = st.selectbox(
        "Select scenario",
        available_scenarios
    )


    selected_global = global_scenarios[
        global_scenarios["Scenario"] == selected_scenario
    ].copy()

    selected_global = selected_global.sort_values("Year")


    # ========================================================
    # GLOBAL KPI VALUES
    # ========================================================

    latest_projection = selected_global.iloc[-1]

    projection_2030 = selected_global[
        selected_global["Year"] == 2030
    ]

    projection_2040 = selected_global[
        selected_global["Year"] == 2040
    ]

    projection_2050 = selected_global[
        selected_global["Year"] == 2050
    ]


    c1, c2, c3, c4 = st.columns(4)

    with c1:

        if not projection_2030.empty:

            st.metric(
                "2030",
                f'{projection_2030["Projected_Emissions_Mt"].iloc[0]:,.0f} Mt'
            )

    with c2:

        if not projection_2040.empty:

            st.metric(
                "2040",
                f'{projection_2040["Projected_Emissions_Mt"].iloc[0]:,.0f} Mt'
            )

    with c3:

        if not projection_2050.empty:

            st.metric(
                "2050",
                f'{projection_2050["Projected_Emissions_Mt"].iloc[0]:,.0f} Mt'
            )

    with c4:

        if "Annual_Growth_Rate" in selected_global.columns:

            growth_2050 = projection_2050[
                "Annual_Growth_Rate"
            ].iloc[0]

            st.metric(
                "2050 Growth Rate",
                f"{growth_2050 * 100:.2f}%"
            )


    # ========================================================
    # GLOBAL PROJECTION CHART
    # ========================================================

    st.subheader("3. Global Emissions Projection")

    fig_global = go.Figure()

    for scenario_name in scenario_order:

        scenario_data = global_scenarios[
            global_scenarios["Scenario"] == scenario_name
        ].sort_values("Year")

        if scenario_data.empty:
            continue

        fig_global.add_trace(
            go.Scatter(
                x=scenario_data["Year"],
                y=scenario_data["Projected_Emissions_Mt"],
                mode="lines",
                name=scenario_name
            )
        )


    # Historical line

    historical_path = (
        BASE_DIR
        / "outputs"
        / "global_annual_analysis.csv"
    )

    if historical_path.exists():

        historical_data = pd.read_csv(
            historical_path
        )

        if {
            "Year",
            "Emissions_Mt"
        }.issubset(historical_data.columns):

            fig_global.add_trace(
                go.Scatter(
                    x=historical_data["Year"],
                    y=historical_data["Emissions_Mt"],
                    mode="lines",
                    name="Historical",
                    line=dict(
                        dash="dot"
                    )
                )
            )


    fig_global.update_layout(
        title="Historical Emissions and Scenario Projections",
        xaxis_title="Year",
        yaxis_title="Emissions (Mt)",
        hovermode="x unified"
    )

    st.plotly_chart(
        fig_global,
        use_container_width=True
    )


    # ========================================================
    # KEY YEAR COMPARISON
    # ========================================================

    st.subheader("4. Scenario Comparison at Key Years")

    comparison_data = (
        global_scenarios[
            global_scenarios["Year"].isin(
                [2030, 2040, 2050]
            )
        ]
        .pivot_table(
            index="Scenario",
            columns="Year",
            values="Projected_Emissions_Mt",
            aggfunc="sum"
        )
        .reset_index()
    )

    st.dataframe(
        comparison_data,
        use_container_width=True,
        hide_index=True
    )


    # ========================================================
    # DIFFERENCE FROM BASELINE
    # ========================================================

    st.subheader("5. Difference Relative to Baseline")

    baseline_data = global_scenarios[
        global_scenarios["Scenario"] == "Baseline"
    ][
        ["Year", "Projected_Emissions_Mt"]
    ].rename(
        columns={
            "Projected_Emissions_Mt":
            "Baseline_Emissions_Mt"
        }
    )


    scenario_difference = global_scenarios.merge(
        baseline_data,
        on="Year",
        how="left"
    )


    scenario_difference[
        "Difference_vs_Baseline_Mt"
    ] = (
        scenario_difference["Projected_Emissions_Mt"]
        - scenario_difference["Baseline_Emissions_Mt"]
    )


    scenario_difference[
        "Difference_vs_Baseline_%"
    ] = (
        scenario_difference[
            "Difference_vs_Baseline_Mt"
        ]
        / scenario_difference[
            "Baseline_Emissions_Mt"
        ].replace(0, np.nan)
        * 100
    )


    selected_difference = scenario_difference[
        scenario_difference["Scenario"] == selected_scenario
    ].copy()


    fig_difference = go.Figure()

    fig_difference.add_trace(
        go.Scatter(
            x=selected_difference["Year"],
            y=selected_difference[
                "Difference_vs_Baseline_Mt"
            ],
            mode="lines",
            name="Difference"
        )
    )

    fig_difference.add_hline(
        y=0,
        line_dash="dash"
    )

    fig_difference.update_layout(
        title=f"{selected_scenario}: Difference vs Baseline",
        xaxis_title="Year",
        yaxis_title="Difference (Mt)"
    )

    st.plotly_chart(
        fig_difference,
        use_container_width=True
    )


    # ========================================================
    # COUNTRY SCENARIO ANALYSIS
    # ========================================================

    if not country_scenarios.empty:

        st.subheader("6. Country Scenario Analysis")

        if "Name" in country_scenarios.columns:

            countries = sorted(
                country_scenarios["Name"]
                .dropna()
                .astype(str)
                .unique()
            )

            selected_country = st.selectbox(
                "Select country",
                countries
            )

            selected_country_data = country_scenarios[
                country_scenarios["Name"].astype(str)
                == selected_country
            ].copy()


            if not selected_country_data.empty:

                st.dataframe(
                    selected_country_data,
                    use_container_width=True,
                    hide_index=True
                )


                # Try to identify scenario and 2050 columns

                scenario_column = None

                for col in [
                    "Scenario",
                    "scenario"
                ]:

                    if col in selected_country_data.columns:

                        scenario_column = col
                        break


                value_column = None

                for col in [
                    "Projected_Emissions_Mt",
                    "Emissions_2050_Mt",
                    "Projected_2050_Mt"
                ]:

                    if col in selected_country_data.columns:

                        value_column = col
                        break


                if (
                    scenario_column is not None
                    and value_column is not None
                ):

                    fig_country = px.bar(
                        selected_country_data,
                        x=scenario_column,
                        y=value_column,
                        title=f"{selected_country} — 2050 Scenario Comparison"
                    )

                    fig_country.update_layout(
                        xaxis_title="Scenario",
                        yaxis_title="Emissions (Mt)"
                    )

                    st.plotly_chart(
                        fig_country,
                        use_container_width=True
                    )


    # ========================================================
    # SECTOR SCENARIO ANALYSIS
    # ========================================================

    if not sector_scenarios.empty:

        st.subheader("7. Sector Scenario Analysis")

        sector_column = None

        for col in [
            "ipcc_code_2006_for_standard_report_name",
            "Sector",
            "sector"
        ]:

            if col in sector_scenarios.columns:

                sector_column = col
                break


        if sector_column is not None:

            sectors = sorted(
                sector_scenarios[sector_column]
                .dropna()
                .astype(str)
                .unique()
            )

            selected_sector = st.selectbox(
                "Select sector",
                sectors
            )

            selected_sector_data = sector_scenarios[
                sector_scenarios[sector_column].astype(str)
                == selected_sector
            ].copy()


            st.dataframe(
                selected_sector_data,
                use_container_width=True,
                hide_index=True
            )


            scenario_column = None

            for col in [
                "Scenario",
                "scenario"
            ]:

                if col in selected_sector_data.columns:

                    scenario_column = col
                    break


            value_column = None

            for col in [
                "Projected_Emissions_Mt",
                "Emissions_2050_Mt",
                "Projected_2050_Mt"
            ]:

                if col in selected_sector_data.columns:

                    value_column = col
                    break


            if (
                scenario_column is not None
                and value_column is not None
            ):

                fig_sector = px.bar(
                    selected_sector_data,
                    x=scenario_column,
                    y=value_column,
                    title=f"{selected_sector} — 2050 Scenario Comparison"
                )

                fig_sector.update_layout(
                    xaxis_title="Scenario",
                    yaxis_title="Emissions (Mt)"
                )

                st.plotly_chart(
                    fig_sector,
                    use_container_width=True
                )


    # ========================================================
    # COUNTRY × SECTOR SCENARIO
    # ========================================================

    if not country_sector_scenarios.empty:

        st.subheader("8. Country × Sector Scenario Explorer")

        cs_data = country_sector_scenarios.copy()

        country_column = None

        for col in [
            "Name",
            "Country",
            "Country_Name"
        ]:

            if col in cs_data.columns:

                country_column = col
                break


        sector_column = None

        for col in [
            "ipcc_code_2006_for_standard_report_name",
            "Sector",
            "sector"
        ]:

            if col in cs_data.columns:

                sector_column = col
                break


        if country_column is not None:

            country_options = sorted(
                cs_data[country_column]
                .dropna()
                .astype(str)
                .unique()
            )

            explorer_country = st.selectbox(
                "Country × sector country",
                country_options
            )

            filtered_cs = cs_data[
                cs_data[country_column].astype(str)
                == explorer_country
            ].copy()


            if sector_column is not None:

                sector_options = sorted(
                    filtered_cs[sector_column]
                    .dropna()
                    .astype(str)
                    .unique()
                )

                if sector_options:

                    explorer_sector = st.selectbox(
                        "Country × sector sector",
                        sector_options
                    )

                    filtered_cs = filtered_cs[
                        filtered_cs[sector_column].astype(str)
                        == explorer_sector
                    ]


            st.dataframe(
                filtered_cs,
                use_container_width=True,
                hide_index=True
            )


    # ========================================================
    # METHODOLOGY
    # ========================================================

    with st.expander("Scenario Methodology"):

        st.markdown("""
        ### Historical baseline

        The projection engine begins from the latest historical
        emissions observation.

        ### Global growth assumption

        The baseline uses the recent historical growth trajectory.

        ### Scenario adjustments

        **Baseline**

        Recent historical growth trend continues.

        **Moderate Reduction**

        Annual growth rate is reduced by 1 percentage point.

        **Accelerated Reduction**

        Annual growth rate is reduced by 2 percentage points.

        ### Important limitation

        These scenarios are not forecasts of what will necessarily happen.

        They are transparent what-if trajectories generated from explicit
        assumptions.

        The scenario engine is intentionally separated from the ML
        validation model.
        """)


    st.warning(
        "Scenario outputs are assumption-sensitive. "
        "Changing the underlying growth assumptions can materially "
        "change the 2030–2050 trajectory."
    )
'''

    # ========================================================
    # WRITE UPDATED APP
    # ========================================================

    app_code = (
        app_code[:start]
        + scenario_page
        + app_code[end:]
    )

    APP_FILE.write_text(
        app_code,
        encoding="utf-8"
    )

    print("============================================================")
    print("PART 6H — SCENARIO SIMULATOR ADDED")
    print("============================================================")
    print()
    print("Updated:", APP_FILE)
    print()
    print("Included:")
    print("✓ Scenario assumptions")
    print("✓ Baseline")
    print("✓ Moderate Reduction")
    print("✓ Accelerated Reduction")
    print("✓ 2030 / 2040 / 2050 KPIs")
    print("✓ Historical + scenario projection")
    print("✓ Scenario comparison")
    print("✓ Difference vs baseline")
    print("✓ Country scenarios")
    print("✓ Sector scenarios")
    print("✓ Country × sector explorer")
    print("✓ Methodology and limitations")
```

```python
# ============================================================
# PART 6I — MODEL DIAGNOSTICS
# ============================================================

from pathlib import Path

PROJECT_DIR = Path.cwd()
APP_FILE = PROJECT_DIR / "app.py"

app_code = APP_FILE.read_text(encoding="utf-8")

marker = 'elif page == "Model Diagnostics":'

if marker not in app_code:

    print("ERROR: Model Diagnostics section not found.")
    print()
    print("Available pages:")

    for line in app_code.splitlines():
        if "elif page ==" in line:
            print(line)

else:

    start = app_code.index(marker)

    remaining = app_code[start + len(marker):]

    next_marker = remaining.find("\nelif page ==")

    if next_marker == -1:
        end = len(app_code)
    else:
        end = start + len(marker) + next_marker

    # ========================================================
    # MODEL DIAGNOSTICS PAGE
    # ========================================================

    diagnostics_page = r'''
elif page == "Model Diagnostics":

    st.title("🔬 Model Diagnostics")

    st.caption(
        "Detailed diagnostic analysis of the historical ML validation."
    )

    # ========================================================
    # LOAD DIAGNOSTIC ARTIFACTS
    # ========================================================

    @st.cache_data
    def load_diagnostic_predictions():

        path = BASE_DIR / "outputs" / "ridge_predictions.csv"

        if not path.exists():
            return pd.DataFrame()

        return pd.read_csv(path)


    @st.cache_data
    def load_model_comparison_diagnostics():

        path = BASE_DIR / "outputs" / "model_comparison.csv"

        if not path.exists():
            return pd.DataFrame()

        return pd.read_csv(path)


    @st.cache_data
    def load_year_error_diagnostics():

        path = BASE_DIR / "outputs" / "year_wise_model_error.csv"

        if not path.exists():
            return pd.DataFrame()

        return pd.read_csv(path)


    @st.cache_data
    def load_country_error_diagnostics():

        path = BASE_DIR / "outputs" / "country_wise_model_error.csv"

        if not path.exists():
            return pd.DataFrame()

        return pd.read_csv(path)


    @st.cache_data
    def load_sector_error_diagnostics():

        path = BASE_DIR / "outputs" / "sector_wise_model_error.csv"

        if not path.exists():
            return pd.DataFrame()

        return pd.read_csv(path)


    predictions_diag = load_diagnostic_predictions()
    model_comparison_diag = load_model_comparison_diagnostics()
    year_error_diag = load_year_error_diagnostics()
    country_error_diag = load_country_error_diagnostics()
    sector_error_diag = load_sector_error_diagnostics()


    # ========================================================
    # VALIDATION
    # ========================================================

    if predictions_diag.empty:

        st.error(
            "ridge_predictions.csv is missing or empty."
        )

        st.stop()


    # ========================================================
    # VALIDATION CONTEXT
    # ========================================================

    st.info(
        "Diagnostic results are based on the chronological validation "
        "period 2020–2023. They should be interpreted as validation "
        "evidence rather than proof of future performance."
    )


    # ========================================================
    # PREPARE RESIDUALS
    # ========================================================

    if {
        "Actual_Mt",
        "Predicted_Mt"
    }.issubset(predictions_diag.columns):

        predictions_diag = predictions_diag.copy()

        predictions_diag["Residual_Mt"] = (
            predictions_diag["Actual_Mt"]
            - predictions_diag["Predicted_Mt"]
        )

        predictions_diag["Absolute_Error_Mt"] = (
            predictions_diag["Residual_Mt"].abs()
        )

        predictions_diag["Squared_Error_Mt2"] = (
            predictions_diag["Residual_Mt"] ** 2
        )


    # ========================================================
    # TOP LEVEL DIAGNOSTIC KPIs
    # ========================================================

    st.subheader("1. Diagnostic Summary")

    if "Residual_Mt" in predictions_diag.columns:

        mean_residual = predictions_diag["Residual_Mt"].mean()

        median_abs_error = (
            predictions_diag["Absolute_Error_Mt"].median()
        )

        max_abs_error = (
            predictions_diag["Absolute_Error_Mt"].max()
        )

        rmse_diag = np.sqrt(
            predictions_diag["Squared_Error_Mt2"].mean()
        )

        c1, c2, c3, c4 = st.columns(4)

        with c1:
            st.metric(
                "Mean Residual",
                f"{mean_residual:.3f} Mt"
            )

        with c2:
            st.metric(
                "Median Absolute Error",
                f"{median_abs_error:.3f} Mt"
            )

        with c3:
            st.metric(
                "RMSE",
                f"{rmse_diag:.3f} Mt"
            )

        with c4:
            st.metric(
                "Maximum Absolute Error",
                f"{max_abs_error:.3f} Mt"
            )


    # ========================================================
    # ACTUAL VS PREDICTED SCATTER
    # ========================================================

    st.subheader("2. Actual vs Predicted")

    if {
        "Actual_Mt",
        "Predicted_Mt"
    }.issubset(predictions_diag.columns):

        fig_scatter = px.scatter(
            predictions_diag,
            x="Actual_Mt",
            y="Predicted_Mt",
            opacity=0.55,
            title="Actual vs Predicted Emissions"
        )

        min_value = min(
            predictions_diag["Actual_Mt"].min(),
            predictions_diag["Predicted_Mt"].min()
        )

        max_value = max(
            predictions_diag["Actual_Mt"].max(),
            predictions_diag["Predicted_Mt"].max()
        )

        fig_scatter.add_trace(
            go.Scatter(
                x=[min_value, max_value],
                y=[min_value, max_value],
                mode="lines",
                name="Perfect Prediction"
            )
        )

        fig_scatter.update_layout(
            xaxis_title="Actual Emissions (Mt)",
            yaxis_title="Predicted Emissions (Mt)"
        )

        st.plotly_chart(
            fig_scatter,
            use_container_width=True
        )


    # ========================================================
    # RESIDUAL DISTRIBUTION
    # ========================================================

    st.subheader("3. Residual Distribution")

    if "Residual_Mt" in predictions_diag.columns:

        fig_residual = px.histogram(
            predictions_diag,
            x="Residual_Mt",
            nbins=60,
            title="Distribution of Prediction Residuals"
        )

        fig_residual.add_vline(
            x=0,
            line_dash="dash"
        )

        fig_residual.update_layout(
            xaxis_title="Residual (Actual − Predicted)",
            yaxis_title="Frequency"
        )

        st.plotly_chart(
            fig_residual,
            use_container_width=True
        )


    # ========================================================
    # RESIDUALS VS PREDICTED
    # ========================================================

    st.subheader("4. Residuals vs Predicted")

    if {
        "Residual_Mt",
        "Predicted_Mt"
    }.issubset(predictions_diag.columns):

        fig_residual_pred = px.scatter(
            predictions_diag,
            x="Predicted_Mt",
            y="Residual_Mt",
            opacity=0.5,
            title="Residuals vs Predicted Emissions"
        )

        fig_residual_pred.add_hline(
            y=0,
            line_dash="dash"
        )

        fig_residual_pred.update_layout(
            xaxis_title="Predicted Emissions (Mt)",
            yaxis_title="Residual (Mt)"
        )

        st.plotly_chart(
            fig_residual_pred,
            use_container_width=True
        )


    # ========================================================
    # ERROR BY YEAR
    # ========================================================

    st.subheader("5. Error by Validation Year")

    if not year_error_diag.empty:

        st.dataframe(
            year_error_diag,
            use_container_width=True,
            hide_index=True
        )

        metric_options = [
            col
            for col in [
                "MAE_Mt",
                "RMSE_Mt",
                "R2"
            ]
            if col in year_error_diag.columns
        ]

        if metric_options:

            selected_year_metric = st.selectbox(
                "Diagnostic metric by year",
                metric_options
            )

            fig_year_diag = px.line(
                year_error_diag,
                x="Year",
                y=selected_year_metric,
                markers=True,
                title=f"{selected_year_metric} by Year"
            )

            fig_year_diag.update_layout(
                xaxis_title="Year",
                yaxis_title=selected_year_metric
            )

            st.plotly_chart(
                fig_year_diag,
                use_container_width=True
            )


    # ========================================================
    # COUNTRY ERROR
    # ========================================================

    st.subheader("6. Country-Level Error")

    if not country_error_diag.empty:

        st.dataframe(
            country_error_diag,
            use_container_width=True,
            hide_index=True
        )

        country_metric_options = [
            col
            for col in [
                "MAE_Mt",
                "RMSE_Mt",
                "R2",
                "Mean_Error_Mt"
            ]
            if col in country_error_diag.columns
        ]

        if country_metric_options:

            selected_country_metric = st.selectbox(
                "Country diagnostic metric",
                country_metric_options
            )

            country_plot = (
                country_error_diag
                .sort_values(
                    selected_country_metric,
                    ascending=False
                )
                .head(20)
                .copy()
            )

            name_column = (
                "Name"
                if "Name" in country_plot.columns
                else "Country_code_A3"
            )

            fig_country_diag = px.bar(
                country_plot,
                x=selected_country_metric,
                y=name_column,
                orientation="h",
                title=f"Top 20 Countries by {selected_country_metric}"
            )

            fig_country_diag.update_layout(
                yaxis_title="Country",
                xaxis_title=selected_country_metric
            )

            st.plotly_chart(
                fig_country_diag,
                use_container_width=True
            )


    # ========================================================
    # SECTOR ERROR
    # ========================================================

    st.subheader("7. Sector-Level Error")

    if not sector_error_diag.empty:

        st.dataframe(
            sector_error_diag,
            use_container_width=True,
            hide_index=True
        )

        sector_metric_options = [
            col
            for col in [
                "MAE_Mt",
                "RMSE_Mt",
                "R2",
                "Mean_Error_Mt"
            ]
            if col in sector_error_diag.columns
        ]

        if sector_metric_options:

            selected_sector_metric = st.selectbox(
                "Sector diagnostic metric",
                sector_metric_options
            )

            sector_plot = (
                sector_error_diag
                .sort_values(
                    selected_sector_metric,
                    ascending=False
                )
                .head(20)
                .copy()
            )

            sector_name_column = None

            for col in [
                "ipcc_code_2006_for_standard_report_name",
                "Sector",
                "sector"
            ]:

                if col in sector_plot.columns:

                    sector_name_column = col
                    break


            if sector_name_column is not None:

                fig_sector_diag = px.bar(
                    sector_plot,
                    x=selected_sector_metric,
                    y=sector_name_column,
                    orientation="h",
                    title=f"Top 20 Sectors by {selected_sector_metric}"
                )

                fig_sector_diag.update_layout(
                    yaxis_title="Sector",
                    xaxis_title=selected_sector_metric
                )

                st.plotly_chart(
                    fig_sector_diag,
                    use_container_width=True
                )


    # ========================================================
    # LARGEST ERRORS
    # ========================================================

    st.subheader("8. Largest Prediction Errors")

    if "Absolute_Error_Mt" in predictions_diag.columns:

        largest_errors = (
            predictions_diag
            .sort_values(
                "Absolute_Error_Mt",
                ascending=False
            )
            .head(25)
            .copy()
        )

        display_columns = [
            col
            for col in [
                "Year",
                "Country_code_A3",
                "Name",
                "ipcc_code_2006_for_standard_report_name",
                "Actual_Mt",
                "Predicted_Mt",
                "Residual_Mt",
                "Absolute_Error_Mt"
            ]
            if col in largest_errors.columns
        ]

        st.dataframe(
            largest_errors[display_columns],
            use_container_width=True,
            hide_index=True
        )


    # ========================================================
    # MODEL COMPARISON
    # ========================================================

    st.subheader("9. Model Comparison")

    if not model_comparison_diag.empty:

        st.dataframe(
            model_comparison_diag,
            use_container_width=True,
            hide_index=True
        )

        if {
            "Model",
            "RMSE_Mt"
        }.issubset(model_comparison_diag.columns):

            fig_models = px.bar(
                model_comparison_diag,
                x="Model",
                y="RMSE_Mt",
                title="Validation RMSE by Model"
            )

            fig_models.update_layout(
                xaxis_title="Model",
                yaxis_title="RMSE (Mt)"
            )

            st.plotly_chart(
                fig_models,
                use_container_width=True
            )


    # ========================================================
    # DIAGNOSTIC INTERPRETATION
    # ========================================================

    with st.expander("How to Read These Diagnostics"):

        st.markdown("""
        ### Actual vs Predicted

        Points closer to the diagonal indicate smaller prediction errors.

        ### Residual Distribution

        Residuals are calculated as:

        **Residual = Actual − Predicted**

        Values above zero indicate under-prediction.

        Values below zero indicate over-prediction.

        ### Residuals vs Predicted

        This plot helps inspect whether errors change systematically
        with the predicted emission level.

        ### Error by Year

        This reveals whether validation performance changes across
        the 2020–2023 validation period.

        ### Country-Level Error

        This identifies where prediction errors are larger or smaller
        across countries.

        ### Sector-Level Error

        This reveals whether model performance differs across sectors.

        ### Largest Errors

        These observations deserve additional investigation because
        aggregate metrics can hide individual large errors.

        ### Important limitation

        These diagnostics describe model behavior on the saved
        validation dataset. They do not establish causal relationships
        between the features and emissions.
        """)


    # ========================================================
    # DATA DOWNLOAD
    # ========================================================

    st.subheader("10. Diagnostic Dataset")

    csv_data = predictions_diag.to_csv(
        index=False
    ).encode("utf-8")

    st.download_button(
        label="Download Diagnostic Predictions CSV",
        data=csv_data,
        file_name="ridge_diagnostic_predictions.csv",
        mime="text/csv"
    )
'''

    # ========================================================
    # WRITE APP
    # ========================================================

    app_code = (
        app_code[:start]
        + diagnostics_page
        + app_code[end:]
    )

    APP_FILE.write_text(
        app_code,
        encoding="utf-8"
    )

    print("============================================================")
    print("PART 6I — MODEL DIAGNOSTICS ADDED")
    print("============================================================")
    print()
    print("Included:")
    print("✓ Diagnostic KPIs")
    print("✓ Actual vs predicted scatter")
    print("✓ Residual distribution")
    print("✓ Residual vs predicted")
    print("✓ Year-level diagnostics")
    print("✓ Country-level diagnostics")
    print("✓ Sector-level diagnostics")
    print("✓ Largest prediction errors")
    print("✓ Model comparison")
    print("✓ Diagnostic interpretation")
    print("✓ CSV download")
```

```python
# ============================================================
# PART 6J — DATA EXPLORER
# ============================================================

from pathlib import Path

PROJECT_DIR = Path.cwd()
APP_FILE = PROJECT_DIR / "app.py"

app_code = APP_FILE.read_text(encoding="utf-8")

marker = 'elif page == "Data Explorer":'

if marker not in app_code:

    print("ERROR: Data Explorer section not found.")
    print()
    print("Available pages:")

    for line in app_code.splitlines():
        if "elif page ==" in line:
            print(line)

else:

    start = app_code.index(marker)

    remaining = app_code[start + len(marker):]

    next_marker = remaining.find("\nelif page ==")

    if next_marker == -1:
        end = len(app_code)
    else:
        end = start + len(marker) + next_marker

    # ========================================================
    # DATA EXPLORER PAGE
    # ========================================================

    explorer_page = r'''
elif page == "Data Explorer":

    st.title("🔎 Data Explorer")

    st.caption(
        "Interactive exploration of the processed IEA–EDGAR emissions dataset."
    )

    # ========================================================
    # LOAD DATA
    # ========================================================

    @st.cache_data
    def load_explorer_data():

        path = BASE_DIR / "data" / "annual_emissions.csv"

        if not path.exists():
            return pd.DataFrame()

        data = pd.read_csv(path)

        if "Emissions_Gg" in data.columns:

            data["Emissions_Mt"] = (
                pd.to_numeric(
                    data["Emissions_Gg"],
                    errors="coerce"
                ) / 1000
            )

        if "Year" in data.columns:

            data["Year"] = pd.to_numeric(
                data["Year"],
                errors="coerce"
            )

        return data


    explorer_data = load_explorer_data()

    # ========================================================
    # VALIDATION
    # ========================================================

    if explorer_data.empty:

        st.error(
            "Annual emissions dataset could not be loaded."
        )

        st.stop()


    required_columns = [
        "Year",
        "Emissions_Mt"
    ]

    missing_columns = [
        col
        for col in required_columns
        if col not in explorer_data.columns
    ]

    if missing_columns:

        st.error(
            "Required columns missing: "
            + ", ".join(missing_columns)
        )

        st.stop()


    # ========================================================
    # DATASET OVERVIEW
    # ========================================================

    st.subheader("1. Dataset Overview")

    min_year = int(
        explorer_data["Year"].min()
    )

    max_year = int(
        explorer_data["Year"].max()
    )

    total_records = len(explorer_data)

    countries_count = (
        explorer_data["Country_code_A3"]
        .nunique()
        if "Country_code_A3" in explorer_data.columns
        else 0
    )

    sectors_count = (
        explorer_data[
            "ipcc_code_2006_for_standard_report_name"
        ].nunique()
        if "ipcc_code_2006_for_standard_report_name"
        in explorer_data.columns
        else 0
    )


    c1, c2, c3, c4 = st.columns(4)

    with c1:
        st.metric(
            "Records",
            f"{total_records:,}"
        )

    with c2:
        st.metric(
            "Year Range",
            f"{min_year}–{max_year}"
        )

    with c3:
        st.metric(
            "Countries",
            f"{countries_count:,}"
        )

    with c4:
        st.metric(
            "Sectors",
            f"{sectors_count:,}"
        )


    # ========================================================
    # FILTERS
    # ========================================================

    st.subheader("2. Filters")

    filter_col1, filter_col2 = st.columns(2)

    with filter_col1:

        selected_year_range = st.slider(
            "Year range",
            min_value=min_year,
            max_value=max_year,
            value=(min_year, max_year),
            step=1
        )


    with filter_col2:

        if "Country_code_A3" in explorer_data.columns:

            country_options = sorted(
                explorer_data["Country_code_A3"]
                .dropna()
                .astype(str)
                .unique()
            )

            selected_countries = st.multiselect(
                "Country",
                country_options,
                default=[]
            )

        else:

            selected_countries = []


    if (
        "ipcc_code_2006_for_standard_report_name"
        in explorer_data.columns
    ):

        sector_options = sorted(
            explorer_data[
                "ipcc_code_2006_for_standard_report_name"
            ]
            .dropna()
            .astype(str)
            .unique()
        )

        selected_sectors = st.multiselect(
            "Sector",
            sector_options,
            default=[]
        )

    else:

        selected_sectors = []


    # ========================================================
    # EMISSIONS FILTER
    # ========================================================

    min_emissions = float(
        explorer_data["Emissions_Mt"]
        .min()
    )

    max_emissions = float(
        explorer_data["Emissions_Mt"]
        .max()
    )


    if min_emissions < max_emissions:

        selected_emissions_range = st.slider(
            "Annual emissions range (Mt)",
            min_value=float(min_emissions),
            max_value=float(max_emissions),
            value=(
                float(min_emissions),
                float(max_emissions)
            )
        )

    else:

        selected_emissions_range = (
            min_emissions,
            max_emissions
        )


    # ========================================================
    # APPLY FILTERS
    # ========================================================

    filtered_data = explorer_data[
        explorer_data["Year"].between(
            selected_year_range[0],
            selected_year_range[1]
        )
    ].copy()


    if selected_countries:

        filtered_data = filtered_data[
            filtered_data["Country_code_A3"]
            .astype(str)
            .isin(selected_countries)
        ]


    if selected_sectors:

        filtered_data = filtered_data[
            filtered_data[
                "ipcc_code_2006_for_standard_report_name"
            ]
            .astype(str)
            .isin(selected_sectors)
        ]


    filtered_data = filtered_data[
        filtered_data["Emissions_Mt"].between(
            selected_emissions_range[0],
            selected_emissions_range[1]
        )
    ]


    # ========================================================
    # FILTERED KPI
    # ========================================================

    st.subheader("3. Filtered Dataset")

    filtered_records = len(filtered_data)

    filtered_emissions = (
        filtered_data["Emissions_Mt"].sum()
        if not filtered_data.empty
        else 0
    )


    c1, c2, c3 = st.columns(3)

    with c1:

        st.metric(
            "Filtered Records",
            f"{filtered_records:,}"
        )

    with c2:

        st.metric(
            "Total Emissions",
            f"{filtered_emissions:,.2f} Mt"
        )

    with c3:

        if total_records > 0:

            percentage_records = (
                filtered_records
                / total_records
                * 100
            )

            st.metric(
                "% of Dataset",
                f"{percentage_records:.2f}%"
            )


    # ========================================================
    # DATA TABLE
    # ========================================================

    if filtered_data.empty:

        st.warning(
            "No records match the selected filters."
        )

    else:

        st.dataframe(
            filtered_data,
            use_container_width=True,
            hide_index=True,
            height=500
        )


    # ========================================================
    # YEARLY PROFILE
    # ========================================================

    st.subheader("4. Filtered Emissions by Year")

    if not filtered_data.empty:

        yearly_profile = (
            filtered_data
            .groupby(
                "Year",
                as_index=False
            )["Emissions_Mt"]
            .sum()
        )


        fig_yearly = px.line(
            yearly_profile,
            x="Year",
            y="Emissions_Mt",
            markers=True,
            title="Filtered Annual Emissions"
        )


        fig_yearly.update_layout(
            xaxis_title="Year",
            yaxis_title="Emissions (Mt)"
        )


        st.plotly_chart(
            fig_yearly,
            use_container_width=True
        )


    # ========================================================
    # COUNTRY PROFILE
    # ========================================================

    if (
        not filtered_data.empty
        and "Name" in filtered_data.columns
    ):

        st.subheader("5. Country Profile")

        country_profile = (
            filtered_data
            .groupby(
                ["Country_code_A3", "Name"],
                as_index=False
            )["Emissions_Mt"]
            .sum()
            .sort_values(
                "Emissions_Mt",
                ascending=False
            )
            .head(20)
        )


        fig_country = px.bar(
            country_profile,
            x="Emissions_Mt",
            y="Name",
            orientation="h",
            title="Top Countries in Filtered Dataset"
        )


        fig_country.update_layout(
            xaxis_title="Emissions (Mt)",
            yaxis_title="Country"
        )


        st.plotly_chart(
            fig_country,
            use_container_width=True
        )


    # ========================================================
    # SECTOR PROFILE
    # ========================================================

    sector_column = (
        "ipcc_code_2006_for_standard_report_name"
        if
        "ipcc_code_2006_for_standard_report_name"
        in filtered_data.columns
        else None
    )


    if (
        not filtered_data.empty
        and sector_column is not None
    ):

        st.subheader("6. Sector Profile")

        sector_profile = (
            filtered_data
            .groupby(
                sector_column,
                as_index=False
            )["Emissions_Mt"]
            .sum()
            .sort_values(
                "Emissions_Mt",
                ascending=False
            )
            .head(20)
        )


        fig_sector = px.bar(
            sector_profile,
            x="Emissions_Mt",
            y=sector_column,
            orientation="h",
            title="Top Sectors in Filtered Dataset"
        )


        fig_sector.update_layout(
            xaxis_title="Emissions (Mt)",
            yaxis_title="Sector"
        )


        st.plotly_chart(
            fig_sector,
            use_container_width=True
        )


    # ========================================================
    # SUMMARY STATISTICS
    # ========================================================

    st.subheader("7. Summary Statistics")

    if not filtered_data.empty:

        numeric_summary = (
            filtered_data[
                ["Emissions_Gg", "Emissions_Mt"]
            ]
            .describe()
            .T
        )


        st.dataframe(
            numeric_summary,
            use_container_width=True
        )


    # ========================================================
    # NULL ANALYSIS
    # ========================================================

    st.subheader("8. Data Quality Snapshot")

    quality_columns = [
        col
        for col in [
            "Year",
            "Country_code_A3",
            "Name",
            "ipcc_code_2006_for_standard_report_name",
            "Emissions_Gg",
            "Emissions_Mt"
        ]
        if col in filtered_data.columns
    ]


    if quality_columns:

        quality_table = pd.DataFrame({
            "Column": quality_columns,
            "Missing Values": [
                filtered_data[col].isna().sum()
                for col in quality_columns
            ],
            "Missing %": [
                filtered_data[col].isna().mean() * 100
                for col in quality_columns
            ],
            "Unique Values": [
                filtered_data[col].nunique(dropna=True)
                for col in quality_columns
            ]
        })


        st.dataframe(
            quality_table,
            use_container_width=True,
            hide_index=True
        )


    # ========================================================
    # DOWNLOAD
    # ========================================================

    st.subheader("9. Export Filtered Data")

    download_data = filtered_data.to_csv(
        index=False
    ).encode("utf-8")


    st.download_button(
        label="Download Filtered CSV",
        data=download_data,
        file_name="filtered_iea_edgar_emissions.csv",
        mime="text/csv"
    )


    # ========================================================
    # DATA DICTIONARY
    # ========================================================

    with st.expander("Data Dictionary"):

        st.markdown("""
        | Field | Description |
        |---|---|
        | Year | Observation year |
        | Country_code_A3 | ISO-style three-letter country code |
        | Name | Country or entity name |
        | ipcc_code_2006_for_standard_report_name | IPCC emissions sector |
        | fossil_bio | Fossil / biological classification |
        | Substance | Emission substance |
        | Emissions_Gg | Annual emissions in gigagrams |
        | Emissions_Mt | Annual emissions converted to million tonnes |

        **Conversion**

        1,000 Gg = 1 Mt

        The explorer uses the processed annual dataset generated
        by the notebook rather than recalculating the raw monthly
        dataset inside Streamlit.
        """)


    st.info(
        "Filters operate on the annual processed dataset. "
        "Use the notebook for the full monthly transformation pipeline."
    )
'''

    # ========================================================
    # WRITE UPDATED APP
    # ========================================================

    app_code = (
        app_code[:start]
        + explorer_page
        + app_code[end:]
    )

    APP_FILE.write_text(
        app_code,
        encoding="utf-8"
    )

    print("============================================================")
    print("PART 6J — DATA EXPLORER ADDED")
    print("============================================================")
    print()
    print("Included:")
    print("✓ Dataset overview")
    print("✓ Year filter")
    print("✓ Country filter")
    print("✓ Sector filter")
    print("✓ Emissions range filter")
    print("✓ Filtered dataset table")
    print("✓ Yearly emissions profile")
    print("✓ Country profile")
    print("✓ Sector profile")
    print("✓ Summary statistics")
    print("✓ Data quality snapshot")
    print("✓ CSV download")
    print("✓ Data dictionary")
```
