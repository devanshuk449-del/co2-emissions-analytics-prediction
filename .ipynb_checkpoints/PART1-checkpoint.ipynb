```python
# IEA–EDGAR CO₂ EMISSIONS INTELLIGENCE PLATFORM
# Step 1: Imports
import os
import warnings
import joblib
import numpy as np
import pandas as pd

import matplotlib.pyplot as plt
import seaborn as sns

from pathlib import Path
from datetime import datetime

from sklearn.compose import ColumnTransformer
from sklearn.pipeline import Pipeline
from sklearn.preprocessing import (
    OneHotEncoder,
    StandardScaler
)
from sklearn.impute import SimpleImputer

from sklearn.linear_model import Ridge
from sklearn.ensemble import (
    RandomForestRegressor,
    GradientBoostingRegressor
)

from sklearn.metrics import (
    mean_absolute_error,
    mean_squared_error,
    r2_score
)

warnings.filterwarnings("ignore")

pd.set_option("display.max_columns", 100)
pd.set_option("display.max_rows", 100)

sns.set_theme(style="whitegrid")

print("Libraries imported successfully.")
```

```python
# PROJECT CONFIGURATION

PROJECT_DIR = Path.cwd()

RAW_FILE = PROJECT_DIR / "IEA_EDGAR_CO2_m_1970_2023.csv"

DATA_DIR = PROJECT_DIR / "data"
MODEL_DIR = PROJECT_DIR / "models"
OUTPUT_DIR = PROJECT_DIR / "outputs"

DATA_DIR.mkdir(exist_ok=True)
MODEL_DIR.mkdir(exist_ok=True)
OUTPUT_DIR.mkdir(exist_ok=True)

MONTH_COLUMNS = [
    "Jan", "Feb", "Mar", "Apr",
    "May", "Jun", "Jul", "Aug",
    "Sep", "Oct", "Nov", "Dec"
]

print("Project directory:")
print(PROJECT_DIR)

print("\nRaw dataset:")
print(RAW_FILE)

print("\nDataset exists:", RAW_FILE.exists())
```

```python

# LOAD RAW DATA


df_raw = pd.read_csv(
    RAW_FILE,
    skiprows=9,
    encoding="latin1",
    low_memory=False
)

print("Dataset loaded successfully.")
print("Rows:", len(df_raw))
print("Columns:", len(df_raw.columns))

df_raw.head()
```

```python
# DATASET OVERVIEW
print("Shape:")
print(df_raw.shape)

print("\nColumns:")
for i, col in enumerate(df_raw.columns, start=1):
    print(f"{i:02d}. {col}")

print("\nData types:")
display(df_raw.dtypes)

print("\nFirst 5 rows:")
display(df_raw.head())
```

```python
# MEMORY USAGE
memory_mb = df_raw.memory_usage(deep=True).sum() / (1024 ** 2)

print(f"Memory usage: {memory_mb:.2f} MB")
```

```python

# DATA QUALITY CHECK


quality = pd.DataFrame({
    "column": df_raw.columns,
    "dtype": df_raw.dtypes.astype(str).values,
    "missing_count": df_raw.isna().sum().values,
    "missing_percentage": (
        df_raw.isna().mean().values * 100
    ).round(3),
    "unique_values": [
        df_raw[col].nunique(dropna=True)
        for col in df_raw.columns
    ]
})

quality.sort_values(
    "missing_percentage",
    ascending=False
)
```

```python
#SUMMARY
df_raw.describe(include="all").T
categorical_columns = [
    "IPCC_annex",
    "C_group_IM24_sh",
    "Country_code_A3",
    "Name",
    "fossil_bio",
    "Substance"
]

for col in categorical_columns:
    print(f"\n{'='*70}")
    print(col)
    print(f"{'='*70}")
    print("Unique:", df_raw[col].nunique())
    print(df_raw[col].dropna().unique()[:20])
```

```python
print("Minimum year:", df_raw["Year"].min())
print("Maximum year:", df_raw["Year"].max())

print("\nYear distribution:")
display(
    df_raw["Year"]
    .value_counts()
    .sort_index()
    .head(20)
)
```

```python
print("Monthly columns:")

for month in MONTH_COLUMNS:
    print(
        f"{month:>3}: "
        f"missing={df_raw[month].isna().sum():,}, "
        f"min={df_raw[month].min()}, "
        f"max={df_raw[month].max()}"
    )
```

```python

# CREATE WORKING COPY


df = df_raw.copy()

# Standardize column names
df.columns = [
    str(col).strip()
    for col in df.columns
]

# Ensure Year is numeric
df["Year"] = pd.to_numeric(
    df["Year"],
    errors="coerce"
)

# Convert monthly values to numeric
for month in MONTH_COLUMNS:
    df[month] = pd.to_numeric(
        df[month],
        errors="coerce"
    )

print("Working dataset prepared.")
print(df.shape)
```

```python

# INVALID VALUE CHECK


invalid_years = df[
    ~df["Year"].between(1970, 2023)
]

print("Invalid year records:", len(invalid_years))

negative_counts = {}

for month in MONTH_COLUMNS:
    negative_counts[month] = (
        (df[month] < 0).sum()
    )

pd.Series(negative_counts)
```

```python

# WIDE → LONG TRANSFORMATION


id_columns = [
    "IPCC_annex",
    "C_group_IM24_sh",
    "Country_code_A3",
    "Name",
    "ipcc_code_2006_for_standard_report",
    "ipcc_code_2006_for_standard_report_name",
    "fossil_bio",
    "Substance",
    "Year"
]

df_monthly = df.melt(
    id_vars=id_columns,
    value_vars=MONTH_COLUMNS,
    var_name="Month_Name",
    value_name="Emissions_Gg"
)

print("Long-format dataset created.")
print("Shape:", df_monthly.shape)

display(df_monthly.head())
```

```python

# MONTH FEATURE
month_mapping = {
    "Jan": 1,
    "Feb": 2,
    "Mar": 3,
    "Apr": 4,
    "May": 5,
    "Jun": 6,
    "Jul": 7,
    "Aug": 8,
    "Sep": 9,
    "Oct": 10,
    "Nov": 11,
    "Dec": 12
}

df_monthly["Month_Num"] = (
    df_monthly["Month_Name"]
    .map(month_mapping)
)

df_monthly.head()
```

```python

# DATE FEATURE


df_monthly["Date"] = pd.to_datetime(
    dict(
        year=df_monthly["Year"],
        month=df_monthly["Month_Num"],
        day=1
    ),
    errors="coerce"
)

df_monthly[
    [
        "Year",
        "Month_Name",
        "Month_Num",
        "Date",
        "Emissions_Gg"
    ]
].head(12)
```

```python
# ============================================================
# TIME FEATURES
# ============================================================

df_monthly["Quarter"] = (
    df_monthly["Month_Num"]
    .sub(1)
    .floordiv(3)
    .add(1)
)

df_monthly["Time_Index"] = (
    (df_monthly["Year"] - df_monthly["Year"].min()) * 12
    + df_monthly["Month_Num"]
)

# Cyclical month encoding
df_monthly["Month_Sin"] = np.sin(
    2 * np.pi * df_monthly["Month_Num"] / 12
)

df_monthly["Month_Cos"] = np.cos(
    2 * np.pi * df_monthly["Month_Num"] / 12
)

df_monthly.head()
```

```python
# ============================================================
# ANNUAL EMISSIONS
# ============================================================

annual_group_columns = [
    "IPCC_annex",
    "C_group_IM24_sh",
    "Country_code_A3",
    "Name",
    "ipcc_code_2006_for_standard_report",
    "ipcc_code_2006_for_standard_report_name",
    "fossil_bio",
    "Substance",
    "Year"
]

df_annual = (
    df_monthly
    .groupby(
        annual_group_columns,
        as_index=False,
        observed=True
    )["Emissions_Gg"]
    .sum()
)

print("Annual dataset:")
print(df_annual.shape)

display(df_annual.head())
```

```python
# ============================================================
# SAVE PROCESSED DATA
# ============================================================

annual_path = DATA_DIR / "annual_emissions.csv"

df_annual.to_csv(
    annual_path,
    index=False
)

print("Saved:")
print(annual_path)
```

```python
# ============================================================
# GLOBAL ANNUAL EMISSIONS
# ============================================================

global_annual = (
    df_annual
    .groupby("Year", as_index=False)["Emissions_Gg"]
    .sum()
)

global_annual["Emissions_Mt"] = (
    global_annual["Emissions_Gg"] / 1_000
)

display(global_annual.head())
```

```python
# ============================================================
# COUNTRY-LEVEL ANNUAL EMISSIONS
# ============================================================

country_annual = (
    df_annual
    .groupby(
        ["Year", "Country_code_A3", "Name"],
        as_index=False,
        observed=True
    )["Emissions_Gg"]
    .sum()
)

country_annual["Emissions_Mt"] = (
    country_annual["Emissions_Gg"] / 1_000
)

print("Country annual dataset created successfully.")
print("Shape:", country_annual.shape)

display(country_annual.head())
```

```python
# ============================================================
# TOP COUNTRIES — LATEST YEAR
# ============================================================

latest_year = country_annual["Year"].max()

top_countries_latest = (
    country_annual[
        country_annual["Year"] == latest_year
    ]
    .sort_values(
        "Emissions_Mt",
        ascending=False
    )
    .head(15)
)

print(f"Top emitting countries in {latest_year}")

display(
    top_countries_latest[
        [
            "Country_code_A3",
            "Name",
            "Emissions_Mt"
        ]
    ]
)
```

```python
# ============================================================
# TOP 15 COUNTRIES
# ============================================================

plt.figure(figsize=(12, 7))

plot_data = top_countries_latest.sort_values(
    "Emissions_Mt"
)

plt.barh(
    plot_data["Name"],
    plot_data["Emissions_Mt"]
)

plt.title(
    f"Top 15 CO₂ Emitting Countries — {latest_year}",
    fontsize=16
)

plt.xlabel("CO₂ Emissions (Mt)")
plt.ylabel("Country")

plt.tight_layout()
plt.show()
```

```python
# ============================================================
# MAJOR COUNTRIES — HISTORICAL TREND
# ============================================================

top_country_names = (
    top_countries_latest["Name"]
    .tolist()
)

country_trend = country_annual[
    country_annual["Name"].isin(top_country_names)
]

plt.figure(figsize=(14, 8))

for country in top_country_names:
    
    temp = country_trend[
        country_trend["Name"] == country
    ]
    
    plt.plot(
        temp["Year"],
        temp["Emissions_Mt"],
        linewidth=2,
        label=country
    )

plt.title(
    "Historical CO₂ Emissions — Major Emitters",
    fontsize=16
)

plt.xlabel("Year")
plt.ylabel("CO₂ Emissions (Mt)")

plt.legend(
    bbox_to_anchor=(1.02, 1),
    loc="upper left"
)

plt.tight_layout()
plt.show()
```

```python
# ============================================================
# COUNTRY YEAR-OVER-YEAR GROWTH
# ============================================================

country_annual = country_annual.sort_values(
    ["Country_code_A3", "Year"]
)

country_annual["YoY_Growth_%"] = (
    country_annual
    .groupby("Country_code_A3")["Emissions_Mt"]
    .pct_change()
    * 100
)

display(
    country_annual.head(20)
)
```

```python
# ============================================================
# LARGEST RECENT INCREASES
# ============================================================

latest_growth = (
    country_annual[
        country_annual["Year"] == latest_year
    ]
    .dropna(subset=["YoY_Growth_%"])
    .sort_values(
        "YoY_Growth_%",
        ascending=False
    )
)

display(
    latest_growth[
        [
            "Name",
            "Emissions_Mt",
            "YoY_Growth_%"
        ]
    ].head(15)
)
```

```python
# ============================================================
# LARGEST RECENT DECREASES
# ============================================================

largest_decreases = (
    country_annual[
        country_annual["Year"] == latest_year
    ]
    .dropna(subset=["YoY_Growth_%"])
    .sort_values(
        "YoY_Growth_%",
        ascending=True
    )
)

display(
    largest_decreases[
        [
            "Name",
            "Emissions_Mt",
            "YoY_Growth_%"
        ]
    ].head(15)
)
```

```python
# ============================================================
# SECTOR-LEVEL ANNUAL EMISSIONS
# ============================================================

sector_annual = (
    df_annual
    .groupby(
        [
            "Year",
            "ipcc_code_2006_for_standard_report",
            "ipcc_code_2006_for_standard_report_name"
        ],
        as_index=False,
        observed=True
    )["Emissions_Gg"]
    .sum()
)

sector_annual["Emissions_Mt"] = (
    sector_annual["Emissions_Gg"] / 1_000
)

print("Sector annual dataset created.")
print("Shape:", sector_annual.shape)

display(sector_annual.head())
```

```python
# ============================================================
# TOP EMITTING SECTORS — LATEST YEAR
# ============================================================

top_sectors_latest = (
    sector_annual[
        sector_annual["Year"] == latest_year
    ]
    .sort_values(
        "Emissions_Mt",
        ascending=False
    )
    .head(15)
)

display(
    top_sectors_latest[
        [
            "ipcc_code_2006_for_standard_report_name",
            "Emissions_Mt"
        ]
    ]
)
```

```python
# ============================================================
# TOP SECTORS VISUALIZATION
# ============================================================

plot_data = top_sectors_latest.sort_values(
    "Emissions_Mt"
)

plt.figure(figsize=(12, 8))

plt.barh(
    plot_data[
        "ipcc_code_2006_for_standard_report_name"
    ],
    plot_data["Emissions_Mt"]
)

plt.title(
    f"Top CO₂ Emitting Sectors — {latest_year}",
    fontsize=16
)

plt.xlabel("CO₂ Emissions (Mt)")
plt.ylabel("Sector")

plt.tight_layout()
plt.show()
```

```python
# ============================================================
# COUNTRY × SECTOR ANNUAL EMISSIONS
# ============================================================

country_sector_annual = (
    df_annual
    .groupby(
        [
            "Year",
            "Country_code_A3",
            "Name",
            "ipcc_code_2006_for_standard_report_name"
        ],
        as_index=False,
        observed=True
    )["Emissions_Gg"]
    .sum()
)

country_sector_annual["Emissions_Mt"] = (
    country_sector_annual["Emissions_Gg"] / 1_000
)

print(
    "Country × Sector dataset:",
    country_sector_annual.shape
)

display(country_sector_annual.head())
```

```python
# ============================================================
# SECTOR CONTRIBUTION WITHIN COUNTRY
# ============================================================

country_sector_annual["Country_Total_Mt"] = (
    country_sector_annual
    .groupby(
        ["Year", "Country_code_A3"]
    )["Emissions_Mt"]
    .transform("sum")
)

country_sector_annual["Sector_Share_%"] = (
    country_sector_annual["Emissions_Mt"]
    / country_sector_annual["Country_Total_Mt"]
    * 100
)

display(
    country_sector_annual.head(20)
)
```

```python
# ============================================================
# COUNTRY SELECTION
# ============================================================

available_countries = (
    country_annual[
        ["Country_code_A3", "Name"]
    ]
    .drop_duplicates()
    .sort_values("Name")
)

display(
    available_countries.head(30)
)
```

```python
# ============================================================
# SELECTED COUNTRY ANALYSIS
# ============================================================

selected_country = "IND"
selected_country_data = country_sector_annual[
    (
        country_sector_annual["Country_code_A3"]
        == selected_country
    )
    &
    (
        country_sector_annual["Year"]
        == latest_year
    )
].sort_values(
    "Emissions_Mt",
    ascending=False
)

display(
    selected_country_data[
        [
            "Name",
            "ipcc_code_2006_for_standard_report_name",
            "Emissions_Mt",
            "Sector_Share_%"
        ]
    ].head(15)
)
```

```python
# ============================================================
# SELECTED COUNTRY — SECTOR COMPOSITION
# ============================================================

plot_data = selected_country_data.head(10)

plt.figure(figsize=(12, 7))

plt.barh(
    plot_data[
        "ipcc_code_2006_for_standard_report_name"
    ][::-1],
    plot_data["Sector_Share_%"][::-1]
)

plt.title(
    f"Sector Contribution — {selected_country} ({latest_year})",
    fontsize=16
)

plt.xlabel("Share of Country Emissions (%)")
plt.ylabel("Sector")

plt.tight_layout()
plt.show()
```

```python
# ============================================================
# COUNTRY SHARE OF GLOBAL EMISSIONS
# ============================================================

global_lookup = global_annual[
    ["Year", "Emissions_Mt"]
].rename(
    columns={
        "Emissions_Mt": "Global_Emissions_Mt"
    }
)

country_annual = country_annual.merge(
    global_lookup,
    on="Year",
    how="left"
)

country_annual["Global_Share_%"] = (
    country_annual["Emissions_Mt"]
    / country_annual["Global_Emissions_Mt"]
    * 100
)

display(
    country_annual[
        [
            "Year",
            "Name",
            "Emissions_Mt",
            "Global_Share_%"
        ]
    ].tail(20)
)
```

```python
# ============================================================
# COUNTRY INTELLIGENCE — LATEST YEAR
# ============================================================

country_intelligence = (
    country_annual[
        country_annual["Year"] == latest_year
    ]
    .copy()
)

country_intelligence["Global_Rank"] = (
    country_intelligence[
        "Emissions_Mt"
    ]
    .rank(
        ascending=False,
        method="min"
    )
)

country_intelligence = (
    country_intelligence
    .sort_values("Global_Rank")
)

country_intelligence[
    [
        "Global_Rank",
        "Country_code_A3",
        "Name",
        "Emissions_Mt",
        "YoY_Growth_%",
        "Global_Share_%"
    ]
].head(20)
```

```python
# ============================================================
# PART 3 — TRENDS, SEASONALITY & ANOMALIES


# ------------------------------------------------------------
# 1. GLOBAL ANNUAL DATA
# ------------------------------------------------------------

global_annual = (
    df_annual
    .groupby("Year", as_index=False)["Emissions_Gg"]
    .sum()
)

global_annual["Emissions_Mt"] = global_annual["Emissions_Gg"] / 1000
global_annual = global_annual.sort_values("Year").reset_index(drop=True)

# Year-over-year growth
global_annual["YoY_Growth_%"] = (
    global_annual["Emissions_Mt"].pct_change() * 100
)

# Rolling trend
global_annual["Rolling_3Y_Mean_Mt"] = (
    global_annual["Emissions_Mt"].rolling(3).mean()
)

# ------------------------------------------------------------
# 2. GLOBAL MONTHLY DATA
# ------------------------------------------------------------

global_monthly = (
    df_monthly
    .groupby(
        ["Year", "Month_Num", "Month_Name", "Date"],
        as_index=False,
        observed=True
    )["Emissions_Gg"]
    .sum()
)

global_monthly["Emissions_Mt"] = (
    global_monthly["Emissions_Gg"] / 1000
)

global_monthly = global_monthly.sort_values("Date").reset_index(drop=True)

# ------------------------------------------------------------
# 3. MONTHLY SEASONALITY
# ------------------------------------------------------------

monthly_seasonality = (
    global_monthly
    .groupby(
        ["Month_Num", "Month_Name"],
        as_index=False,
        observed=True
    )["Emissions_Mt"]
    .mean()
    .rename(columns={"Emissions_Mt": "Average_Emissions_Mt"})
    .sort_values("Month_Num")
)

overall_monthly_average = global_monthly["Emissions_Mt"].mean()

monthly_seasonality["Seasonality_Index"] = (
    monthly_seasonality["Average_Emissions_Mt"]
    / overall_monthly_average
)

monthly_seasonality["Deviation_from_Average_%"] = (
    (monthly_seasonality["Seasonality_Index"] - 1) * 100
)

# ------------------------------------------------------------
# 4. ANNUAL ANOMALY DETECTION
# ------------------------------------------------------------

growth_mean = global_annual["YoY_Growth_%"].dropna().mean()
growth_std = global_annual["YoY_Growth_%"].dropna().std()

global_annual["Growth_Z_Score"] = (
    (global_annual["YoY_Growth_%"] - growth_mean)
    / growth_std
)

global_annual["Growth_Anomaly"] = (
    global_annual["Growth_Z_Score"].abs() >= 2
)

# ------------------------------------------------------------
# 5. SUMMARY
# ------------------------------------------------------------

print("=" * 70)
print("PART 3 COMPLETED")
print("=" * 70)

print(f"\nAnnual records: {len(global_annual):,}")
print(f"Monthly records: {len(global_monthly):,}")

print("\nUnusual annual growth years:")
display(
    global_annual[
        global_annual["Growth_Anomaly"]
    ][
        [
            "Year",
            "Emissions_Mt",
            "YoY_Growth_%",
            "Growth_Z_Score"
        ]
    ]
)

print("\nMonthly seasonality:")
display(monthly_seasonality)
```

```python
# ============================================================
# GLOBAL EMISSIONS + 3-YEAR ROLLING TREND
# ============================================================

plt.figure(figsize=(14, 6))

plt.plot(
    global_annual["Year"],
    global_annual["Emissions_Mt"],
    label="Annual Emissions",
    linewidth=2
)

plt.plot(
    global_annual["Year"],
    global_annual["Rolling_3Y_Mean_Mt"],
    label="3-Year Rolling Average",
    linewidth=2
)

plt.title("Global CO₂ Emissions Trend", fontsize=16)
plt.xlabel("Year")
plt.ylabel("Emissions (Mt)")
plt.legend()

plt.tight_layout()
plt.show()
```

```python
# ============================================================
# MONTHLY SEASONALITY
# ============================================================

plt.figure(figsize=(12, 5))

plt.bar(
    monthly_seasonality["Month_Name"],
    monthly_seasonality["Average_Emissions_Mt"]
)

plt.title("Average Monthly Global CO₂ Emissions", fontsize=16)
plt.xlabel("Month")
plt.ylabel("Average Emissions (Mt)")

plt.tight_layout()
plt.show()
```

```python
# ============================================================
# PART 4A — ML DATASET (ROBUST VERSION)
# ============================================================

import numpy as np
import pandas as pd

# ------------------------------------------------------------
# 1. COUNTRY × SECTOR × YEAR DATA
# ------------------------------------------------------------

country_sector_ml = (
    df_annual
    .groupby(
        [
            "Year",
            "Country_code_A3",
            "Name",
            "ipcc_code_2006_for_standard_report_name"
        ],
        as_index=False,
        observed=True
    )["Emissions_Gg"]
    .sum()
)

country_sector_ml["Emissions_Mt"] = (
    country_sector_ml["Emissions_Gg"] / 1000
)

# ------------------------------------------------------------
# 2. SORT
# ------------------------------------------------------------

group_cols = [
    "Country_code_A3",
    "ipcc_code_2006_for_standard_report_name"
]

country_sector_ml = (
    country_sector_ml
    .sort_values(group_cols + ["Year"])
    .reset_index(drop=True)
)

# ------------------------------------------------------------
# 3. TIME FEATURES
# ------------------------------------------------------------

country_sector_ml["Years_Since_1970"] = (
    country_sector_ml["Year"] - 1970
)

country_sector_ml["Year_Sin"] = np.sin(
    2 * np.pi *
    country_sector_ml["Years_Since_1970"] / 10
)

country_sector_ml["Year_Cos"] = np.cos(
    2 * np.pi *
    country_sector_ml["Years_Since_1970"] / 10
)

# ------------------------------------------------------------
# 4. LAG FEATURES
# ------------------------------------------------------------

country_sector_ml["Lag_1Y_Mt"] = (
    country_sector_ml
    .groupby(group_cols, observed=True)["Emissions_Mt"]
    .shift(1)
)

country_sector_ml["Lag_2Y_Mt"] = (
    country_sector_ml
    .groupby(group_cols, observed=True)["Emissions_Mt"]
    .shift(2)
)

country_sector_ml["Lag_3Y_Mt"] = (
    country_sector_ml
    .groupby(group_cols, observed=True)["Emissions_Mt"]
    .shift(3)
)

# ------------------------------------------------------------
# 5. ROLLING FEATURES
# ------------------------------------------------------------

country_sector_ml["Rolling_3Y_Mean_Mt"] = (
    country_sector_ml
    .groupby(group_cols, observed=True)["Emissions_Mt"]
    .transform(
        lambda x: x.shift(1).rolling(3).mean()
    )
)

country_sector_ml["Rolling_3Y_Std_Mt"] = (
    country_sector_ml
    .groupby(group_cols, observed=True)["Emissions_Mt"]
    .transform(
        lambda x: x.shift(1).rolling(3).std()
    )
)

# ------------------------------------------------------------
# 6. PREVIOUS YEAR GROWTH
# ------------------------------------------------------------

previous_growth = (
    country_sector_ml
    .groupby(group_cols, observed=True)["Emissions_Mt"]
    .pct_change()
    .shift(1)
)

# Convert +inf / -inf to NaN
previous_growth = previous_growth.replace(
    [np.inf, -np.inf],
    np.nan
)

country_sector_ml["Previous_YoY_Growth_%"] = (
    previous_growth * 100
)

# ------------------------------------------------------------
# 7. CLEAN ALL NUMERIC FEATURES
# ------------------------------------------------------------

numeric_check_columns = [
    "Emissions_Mt",
    "Years_Since_1970",
    "Year_Sin",
    "Year_Cos",
    "Lag_1Y_Mt",
    "Lag_2Y_Mt",
    "Lag_3Y_Mt",
    "Rolling_3Y_Mean_Mt",
    "Rolling_3Y_Std_Mt",
    "Previous_YoY_Growth_%"
]

country_sector_ml[numeric_check_columns] = (
    country_sector_ml[numeric_check_columns]
    .replace([np.inf, -np.inf], np.nan)
)

# ------------------------------------------------------------
# 8. REMOVE ROWS WITHOUT REQUIRED HISTORY
# ------------------------------------------------------------

ml_data = country_sector_ml.dropna(
    subset=[
        "Lag_1Y_Mt",
        "Lag_2Y_Mt",
        "Lag_3Y_Mt",
        "Rolling_3Y_Mean_Mt"
    ]
).copy()

# ------------------------------------------------------------
# 9. TIME-AWARE SPLIT
# ------------------------------------------------------------

TRAIN_END_YEAR = 2019

train_data = ml_data[
    ml_data["Year"] <= TRAIN_END_YEAR
].copy()

test_data = ml_data[
    ml_data["Year"] > TRAIN_END_YEAR
].copy()

# ------------------------------------------------------------
# 10. FEATURES
# ------------------------------------------------------------

FEATURES = [
    "Year",
    "Years_Since_1970",
    "Year_Sin",
    "Year_Cos",
    "Lag_1Y_Mt",
    "Lag_2Y_Mt",
    "Lag_3Y_Mt",
    "Rolling_3Y_Mean_Mt",
    "Rolling_3Y_Std_Mt",
    "Previous_YoY_Growth_%",
    "Country_code_A3",
    "ipcc_code_2006_for_standard_report_name"
]

TARGET = "Emissions_Mt"

X_train = train_data[FEATURES].copy()
y_train = train_data[TARGET].copy()

X_test = test_data[FEATURES].copy()
y_test = test_data[TARGET].copy()

# ------------------------------------------------------------
# 11. FINAL INFINITY CHECK
# ------------------------------------------------------------

numeric_train = X_train.select_dtypes(
    include=np.number
)

numeric_test = X_test.select_dtypes(
    include=np.number
)

train_inf_count = np.isinf(
    numeric_train.to_numpy()
).sum()

test_inf_count = np.isinf(
    numeric_test.to_numpy()
).sum()

print("=" * 70)
print("PART 4A — ML DATASET READY")
print("=" * 70)

print(f"\nTotal ML rows : {len(ml_data):,}")
print(f"Training rows : {len(train_data):,}")
print(f"Testing rows  : {len(test_data):,}")

print(
    f"\nTraining period: "
    f"{train_data['Year'].min()} → {train_data['Year'].max()}"
)

print(
    f"Testing period : "
    f"{test_data['Year'].min()} → {test_data['Year'].max()}"
)

print(f"\nInfinite values in X_train: {train_inf_count}")
print(f"Infinite values in X_test : {test_inf_count}")

print(
    f"\nMissing numeric values in X_train: "
    f"{numeric_train.isna().sum().sum():,}"
)

print(
    f"Missing numeric values in X_test: "
    f"{numeric_test.isna().sum().sum():,}"
)

print("\nTarget missing values:")
print("Train:", y_train.isna().sum())
print("Test :", y_test.isna().sum())

print("\nML dataset is ready.")
```

```python
# ============================================================
# PART 4B — PREPROCESSING PIPELINE
# ============================================================

from sklearn.compose import ColumnTransformer
from sklearn.pipeline import Pipeline
from sklearn.preprocessing import (
    OneHotEncoder,
    StandardScaler
)
from sklearn.impute import SimpleImputer

NUMERIC_FEATURES = [
    "Year",
    "Years_Since_1970",
    "Year_Sin",
    "Year_Cos",
    "Lag_1Y_Mt",
    "Lag_2Y_Mt",
    "Lag_3Y_Mt",
    "Rolling_3Y_Mean_Mt",
    "Rolling_3Y_Std_Mt",
    "Previous_YoY_Growth_%"
]

CATEGORICAL_FEATURES = [
    "Country_code_A3",
    "ipcc_code_2006_for_standard_report_name"
]

numeric_pipeline = Pipeline(
    steps=[
        (
            "imputer",
            SimpleImputer(strategy="median")
        ),
        (
            "scaler",
            StandardScaler()
        )
    ]
)

categorical_pipeline = Pipeline(
    steps=[
        (
            "imputer",
            SimpleImputer(strategy="most_frequent")
        ),
        (
            "onehot",
            OneHotEncoder(
                handle_unknown="ignore",
                sparse_output=True
            )
        )
    ]
)

preprocessor = ColumnTransformer(
    transformers=[
        (
            "numeric",
            numeric_pipeline,
            NUMERIC_FEATURES
        ),
        (
            "categorical",
            categorical_pipeline,
            CATEGORICAL_FEATURES
        )
    ]
)

print("=" * 70)
print("PART 4B COMPLETE")
print("=" * 70)

print("\nNumeric features:")
print(NUMERIC_FEATURES)

print("\nCategorical features:")
print(CATEGORICAL_FEATURES)
```

```python
# ============================================================
# PART 4C — MULTI-MODEL TRAINING
# CORRECTED FOR SPARSE / DENSE INPUT
# ============================================================

from sklearn.linear_model import Ridge
from sklearn.ensemble import (
    RandomForestRegressor,
    HistGradientBoostingRegressor
)

from sklearn.metrics import (
    mean_absolute_error,
    mean_squared_error,
    r2_score
)

from sklearn.pipeline import Pipeline
from sklearn.compose import ColumnTransformer
from sklearn.preprocessing import (
    OneHotEncoder,
    StandardScaler
)
from sklearn.impute import SimpleImputer

import numpy as np
import pandas as pd
import time

# ============================================================
# 1. SPARSE PREPROCESSOR
#    Used by Ridge and Random Forest
# ============================================================

numeric_pipeline_sparse = Pipeline(
    steps=[
        (
            "imputer",
            SimpleImputer(strategy="median")
        ),
        (
            "scaler",
            StandardScaler()
        )
    ]
)

categorical_pipeline_sparse = Pipeline(
    steps=[
        (
            "imputer",
            SimpleImputer(strategy="most_frequent")
        ),
        (
            "onehot",
            OneHotEncoder(
                handle_unknown="ignore",
                sparse_output=True
            )
        )
    ]
)

preprocessor_sparse = ColumnTransformer(
    transformers=[
        (
            "numeric",
            numeric_pipeline_sparse,
            NUMERIC_FEATURES
        ),
        (
            "categorical",
            categorical_pipeline_sparse,
            CATEGORICAL_FEATURES
        )
    ]
)

# ============================================================
# 2. DENSE PREPROCESSOR
#    Required by HistGradientBoosting
# ============================================================

numeric_pipeline_dense = Pipeline(
    steps=[
        (
            "imputer",
            SimpleImputer(strategy="median")
        ),
        (
            "scaler",
            StandardScaler()
        )
    ]
)

categorical_pipeline_dense = Pipeline(
    steps=[
        (
            "imputer",
            SimpleImputer(strategy="most_frequent")
        ),
        (
            "onehot",
            OneHotEncoder(
                handle_unknown="ignore",
                sparse_output=False
            )
        )
    ]
)

preprocessor_dense = ColumnTransformer(
    transformers=[
        (
            "numeric",
            numeric_pipeline_dense,
            NUMERIC_FEATURES
        ),
        (
            "categorical",
            categorical_pipeline_dense,
            CATEGORICAL_FEATURES
        )
    ]
)

# ============================================================
# 3. MODELS
# ============================================================

models = {

    "Ridge Regression": (
        preprocessor_sparse,
        Ridge(
            alpha=10.0,
            solver="lsqr"
        )
    ),

    "Random Forest": (
        preprocessor_sparse,
        RandomForestRegressor(
            n_estimators=150,
            max_depth=25,
            min_samples_leaf=2,
            random_state=42,
            n_jobs=-1
        )
    ),

    "Gradient Boosting": (
        preprocessor_dense,
        HistGradientBoostingRegressor(
            max_iter=250,
            learning_rate=0.05,
            max_leaf_nodes=31,
            l2_regularization=1.0,
            random_state=42
        )
    )
}

# ============================================================
# 4. TRAIN
# ============================================================

trained_models = {}
model_predictions = {}
results = []

for model_name, (processor, model) in models.items():

    print("\n" + "=" * 70)
    print(f"TRAINING: {model_name}")
    print("=" * 70)

    start_time = time.time()

    pipeline = Pipeline(
        steps=[
            ("preprocessor", processor),
            ("model", model)
        ]
    )

    pipeline.fit(
        X_train,
        y_train
    )

    predictions = pipeline.predict(
        X_test
    )

    mae = mean_absolute_error(
        y_test,
        predictions
    )

    rmse = np.sqrt(
        mean_squared_error(
            y_test,
            predictions
        )
    )

    r2 = r2_score(
        y_test,
        predictions
    )

    elapsed = time.time() - start_time

    trained_models[model_name] = pipeline
    model_predictions[model_name] = predictions

    results.append({
        "Model": model_name,
        "MAE_Mt": mae,
        "RMSE_Mt": rmse,
        "R2": r2,
        "Training_Time_sec": elapsed
    })

    print(f"MAE  : {mae:.4f} Mt")
    print(f"RMSE : {rmse:.4f} Mt")
    print(f"R²   : {r2:.4f}")
    print(f"Time : {elapsed:.2f} sec")

# ============================================================
# 5. MODEL COMPARISON
# ============================================================

model_results = (
    pd.DataFrame(results)
    .sort_values("RMSE_Mt")
    .reset_index(drop=True)
)

print("\n" + "=" * 70)
print("MODEL COMPARISON")
print("=" * 70)

display(model_results)
```

```python
# ============================================================
# PART 4D — FINAL MODEL COMPARISON
# USING ALREADY COMPLETED MODEL RESULTS
# NO RETRAINING
# ============================================================

# ------------------------------------------------------------
# Gradient Boosting results already obtained
# ------------------------------------------------------------

gb_result = pd.DataFrame([{
    "Model": "Gradient Boosting",
    "MAE_Mt": 1.8990,
    "RMSE_Mt": 37.0276,
    "R2": 0.9201,
    "Training_Time_sec": 22.86
}])

# ------------------------------------------------------------
# Retrieve Ridge + Random Forest results
# ------------------------------------------------------------

existing_results = model_results[
    model_results["Model"].isin([
        "Ridge Regression",
        "Random Forest"
    ])
].copy()

# ------------------------------------------------------------
# Combine all models
# ------------------------------------------------------------

model_results_final = pd.concat(
    [
        existing_results,
        gb_result
    ],
    ignore_index=True
)

# ------------------------------------------------------------
# Sort by RMSE
# ------------------------------------------------------------

model_results_final = (
    model_results_final
    .sort_values("RMSE_Mt")
    .reset_index(drop=True)
)

print("=" * 70)
print("FINAL MODEL COMPARISON")
print("=" * 70)

display(model_results_final)
```

```python
# ============================================================
# PART 4D — FINAL MODEL COMPARISON + RIDGE PREDICTIONS
# ============================================================

# ------------------------------------------------------------
# 1. Reconstruct Gradient Boosting result
# ------------------------------------------------------------

gb_result = pd.DataFrame([{
    "Model": "Gradient Boosting",
    "MAE_Mt": 1.8990,
    "RMSE_Mt": 37.0276,
    "R2": 0.9201,
    "Training_Time_sec": 22.86
}])

# ------------------------------------------------------------
# 2. Retrieve Ridge + Random Forest results
# ------------------------------------------------------------

existing_results = model_results[
    model_results["Model"].isin([
        "Ridge Regression",
        "Random Forest"
    ])
].copy()

# ------------------------------------------------------------
# 3. Combine all three models
# ------------------------------------------------------------

model_results_final = pd.concat(
    [
        existing_results,
        gb_result
    ],
    ignore_index=True
)

# ------------------------------------------------------------
# 4. Sort by test RMSE
# ------------------------------------------------------------

model_results_final = (
    model_results_final
    .sort_values("RMSE_Mt")
    .reset_index(drop=True)
)

print("=" * 70)
print("FINAL MODEL COMPARISON")
print("=" * 70)

display(
    model_results_final.style.format({
        "MAE_Mt": "{:.4f}",
        "RMSE_Mt": "{:.4f}",
        "R2": "{:.4f}",
        "Training_Time_sec": "{:.2f}"
    })
)

# ------------------------------------------------------------
# 5. Ridge is our available prediction model
# ------------------------------------------------------------

best_model_name = "Ridge Regression"

best_predictions = np.asarray(predictions)
actual_values = np.asarray(y_test)

print("\nPrimary prediction model:", best_model_name)
print("Prediction count:", len(best_predictions))
print("Actual count:", len(actual_values))

# ------------------------------------------------------------
# 6. Create prediction dataset
# ------------------------------------------------------------

prediction_results = test_data[
    [
        "Year",
        "Country_code_A3",
        "Name",
        "ipcc_code_2006_for_standard_report_name",
        "Emissions_Mt"
    ]
].copy()

prediction_results = prediction_results.reset_index(drop=True)

# Safety check
if len(prediction_results) != len(best_predictions):
    raise ValueError(
        f"Length mismatch: test_data={len(prediction_results)}, "
        f"predictions={len(best_predictions)}"
    )

prediction_results["Actual_Mt"] = actual_values
prediction_results["Predicted_Mt"] = best_predictions

prediction_results["Error_Mt"] = (
    prediction_results["Actual_Mt"]
    - prediction_results["Predicted_Mt"]
)

prediction_results["Absolute_Error_Mt"] = (
    prediction_results["Error_Mt"].abs()
)

prediction_results["Absolute_Percentage_Error_%"] = (
    prediction_results["Absolute_Error_Mt"]
    / prediction_results["Actual_Mt"].replace(0, np.nan)
    * 100
)

# ------------------------------------------------------------
# 7. Display results
# ------------------------------------------------------------

print("\n" + "=" * 70)
print("RIDGE PREDICTION RESULTS")
print("=" * 70)

display(prediction_results.head(10))

print("\nRows:", len(prediction_results))
print("Columns:", len(prediction_results.columns))
```

```python
# ============================================================
# CHECK AVAILABLE MODEL VARIABLES — SAFE VERSION
# ============================================================

available_model_variables = [
    name for name in list(globals().keys())
    if "ridge" in name.lower()
    or "random" in name.lower()
    or "forest" in name.lower()
    or "gradient" in name.lower()
    or "boost" in name.lower()
    or "model" in name.lower()
]

print("=" * 70)
print("AVAILABLE MODEL VARIABLES")
print("=" * 70)

for name in available_model_variables:
    print(name)
```

```python
# ============================================================
# PART 4G.1 — REBUILD RIDGE PIPELINE
# NO RANDOM FOREST / GRADIENT BOOSTING RETRAINING
# ============================================================

import time

print("=" * 70)
print("REBUILDING RIDGE MODEL FOR EXPLAINABILITY")
print("=" * 70)

ridge_preprocessor = preprocessor_sparse

ridge_model = Pipeline([
    ("preprocessor", ridge_preprocessor),
    ("model", Ridge(
        alpha=10.0,
        solver="lsqr"
    ))
])

start_time = time.time()

ridge_model.fit(X_train, y_train)

ridge_rebuild_time = time.time() - start_time

print(f"Ridge rebuilt successfully in {ridge_rebuild_time:.2f} seconds")
```

```python
# ============================================================
# PART 4G.2 — VERIFY RIDGE PREDICTIONS
# ============================================================

ridge_check_predictions = ridge_model.predict(X_test)

check_mae = mean_absolute_error(
    y_test,
    ridge_check_predictions
)

check_rmse = np.sqrt(
    mean_squared_error(
        y_test,
        ridge_check_predictions
    )
)

check_r2 = r2_score(
    y_test,
    ridge_check_predictions
)

print("=" * 70)
print("RIDGE VERIFICATION")
print("=" * 70)

print(f"MAE  : {check_mae:.4f} Mt")
print(f"RMSE : {check_rmse:.4f} Mt")
print(f"R²   : {check_r2:.4f}")
```

```python
# ============================================================
# PART 4G.3 — RIDGE FEATURE COEFFICIENTS
# ============================================================

fitted_preprocessor = ridge_model.named_steps["preprocessor"]
fitted_ridge = ridge_model.named_steps["model"]

feature_names = fitted_preprocessor.get_feature_names_out()

coefficients = fitted_ridge.coef_

ridge_coefficients = pd.DataFrame({
    "Feature": feature_names,
    "Coefficient": coefficients,
    "Absolute_Coefficient": np.abs(coefficients)
})

ridge_coefficients = (
    ridge_coefficients
    .sort_values(
        "Absolute_Coefficient",
        ascending=False
    )
    .reset_index(drop=True)
)

print("=" * 70)
print("TOP 30 RIDGE FEATURES BY ABSOLUTE COEFFICIENT")
print("=" * 70)

display(
    ridge_coefficients.head(30)
)
```

```python
# ============================================================
# PART 4H.1 — PERMUTATION IMPORTANCE
# ============================================================

from sklearn.inspection import permutation_importance

print("=" * 70)
print("CALCULATING PERMUTATION IMPORTANCE")
print("=" * 70)

permutation_result = permutation_importance(
    ridge_model,
    X_test,
    y_test,
    scoring="neg_root_mean_squared_error",
    n_repeats=5,
    random_state=42,
    n_jobs=1
)

permutation_importance_df = pd.DataFrame({
    "Feature": X_test.columns,
    "Importance_Mean": permutation_result.importances_mean,
    "Importance_Std": permutation_result.importances_std
})

permutation_importance_df = (
    permutation_importance_df
    .sort_values(
        "Importance_Mean",
        ascending=False
    )
    .reset_index(drop=True)
)

print("=" * 70)
print("FEATURE IMPORTANCE — PERMUTATION")
print("=" * 70)

display(permutation_importance_df)
```

```python
# ============================================================
# PART 5A — PROJECTION BASELINE
# ============================================================

print("=" * 70)
print("PART 5A — PROJECTION BASELINE")
print("=" * 70)

# Work from global annual historical emissions
projection_history = global_annual[
    [
        "Year",
        "Emissions_Mt",
        "YoY_Growth_%",
        "Rolling_3Y_Mean_Mt"
    ]
].copy()

projection_history = (
    projection_history
    .sort_values("Year")
    .reset_index(drop=True)
)

# ------------------------------------------------------------
# Historical period
# ------------------------------------------------------------

historical_start_year = int(
    projection_history["Year"].min()
)

historical_end_year = int(
    projection_history["Year"].max()
)

# ------------------------------------------------------------
# Recent historical window
# ------------------------------------------------------------

RECENT_YEARS = 10

recent_history = projection_history.tail(
    RECENT_YEARS
).copy()

recent_growth = (
    recent_history["YoY_Growth_%"]
    .dropna()
)

# Historical CAGR over the recent period
start_emissions = recent_history["Emissions_Mt"].iloc[0]
end_emissions = recent_history["Emissions_Mt"].iloc[-1]

number_of_years = (
    recent_history["Year"].iloc[-1]
    - recent_history["Year"].iloc[0]
)

recent_cagr = (
    (end_emissions / start_emissions)
    ** (1 / number_of_years)
    - 1
)

print(f"Historical period : {historical_start_year}–{historical_end_year}")
print(f"Recent window     : {RECENT_YEARS} years")
print(f"Recent CAGR       : {recent_cagr * 100:.3f}%")
print(f"Average recent YoY: {recent_growth.mean():.3f}%")
print(f"Median recent YoY : {recent_growth.median():.3f}%")
```

```python
# ============================================================
# PART 5B — SCENARIO ASSUMPTIONS
# ============================================================

print("=" * 70)
print("PART 5B — SCENARIO ASSUMPTIONS")
print("=" * 70)

# ------------------------------------------------------------
# Projection horizon
# ------------------------------------------------------------

PROJECTION_END_YEAR = 2050

# ------------------------------------------------------------
# Scenario assumptions
# ------------------------------------------------------------

SCENARIO_ASSUMPTIONS = {
    "Baseline": {
        "annual_adjustment": 0.00,
        "description": "Recent historical growth trend continues"
    },
    "Moderate Reduction": {
        "annual_adjustment": -0.01,
        "description": "Annual growth rate reduced by 1 percentage point"
    },
    "Accelerated Reduction": {
        "annual_adjustment": -0.02,
        "description": "Annual growth rate reduced by 2 percentage points"
    }
}

scenario_table = pd.DataFrame([
    {
        "Scenario": name,
        "Annual_Adjustment": values["annual_adjustment"],
        "Annual_Adjustment_%": values["annual_adjustment"] * 100,
        "Description": values["description"]
    }
    for name, values in SCENARIO_ASSUMPTIONS.items()
])

display(scenario_table)
```

```python
# ============================================================
# PART 5C — GLOBAL SCENARIO PROJECTIONS
# ============================================================

print("=" * 70)
print("PART 5C — GLOBAL SCENARIO PROJECTIONS")
print("=" * 70)

projection_rows = []

last_historical_emissions = (
    projection_history.loc[
        projection_history["Year"] == historical_end_year,
        "Emissions_Mt"
    ].iloc[0]
)

for scenario_name, scenario_info in SCENARIO_ASSUMPTIONS.items():

    scenario_growth = (
        recent_cagr
        + scenario_info["annual_adjustment"]
    )

    previous_emissions = last_historical_emissions

    for year in range(
        historical_end_year + 1,
        PROJECTION_END_YEAR + 1
    ):

        projected_emissions = (
            previous_emissions
            * (1 + scenario_growth)
        )

        projection_rows.append({
            "Year": year,
            "Scenario": scenario_name,
            "Growth_Rate": scenario_growth,
            "Growth_Rate_%": scenario_growth * 100,
            "Projected_Emissions_Mt": projected_emissions
        })

        previous_emissions = projected_emissions

global_projections = pd.DataFrame(projection_rows)

print("Projection rows:", len(global_projections))

display(
    global_projections.head(10)
)
```

```python
# ============================================================
# PART 5D — KEY PROJECTION YEARS
# ============================================================

key_projection_years = [2030, 2040, 2050]

projection_check = (
    global_projections[
        global_projections["Year"].isin(key_projection_years)
    ]
    .pivot(
        index="Scenario",
        columns="Year",
        values="Projected_Emissions_Mt"
    )
    .reset_index()
)

print("=" * 70)
print("KEY PROJECTION YEARS")
print("=" * 70)

display(
    projection_check.style.format({
        2030: "{:,.2f}",
        2040: "{:,.2f}",
        2050: "{:,.2f}"
    })
)
```

```python
# ============================================================
# PART 5E — HISTORICAL + SCENARIO DATA
# ============================================================

historical_plot = projection_history[
    ["Year", "Emissions_Mt"]
].copy()

historical_plot["Scenario"] = "Historical"

scenario_plot = global_projections[
    [
        "Year",
        "Projected_Emissions_Mt",
        "Scenario"
    ]
].copy()

scenario_plot = scenario_plot.rename(
    columns={
        "Projected_Emissions_Mt": "Emissions_Mt"
    }
)

historical_and_scenarios = pd.concat(
    [
        historical_plot,
        scenario_plot
    ],
    ignore_index=True
)

display(
    historical_and_scenarios.tail(10)
)
```

```python
# ============================================================
# GLOBAL EMISSIONS — HISTORICAL + SCENARIOS
# ============================================================

plt.figure(figsize=(13, 7))

plt.plot(
    historical_plot["Year"],
    historical_plot["Emissions_Mt"],
    linewidth=2,
    label="Historical"
)

for scenario in SCENARIO_ASSUMPTIONS.keys():

    scenario_data = global_projections[
        global_projections["Scenario"] == scenario
    ]

    plt.plot(
        scenario_data["Year"],
        scenario_data["Projected_Emissions_Mt"],
        linewidth=2,
        label=scenario
    )

plt.xlabel("Year")
plt.ylabel("CO₂ Emissions (Mt)")
plt.title("Global CO₂ Emissions — Historical and Scenario Projections")

plt.legend()

plt.tight_layout()
plt.show()
```

```python
# ============================================================
# PART 5F — SCENARIO DIFFERENCE ANALYSIS
# ============================================================

baseline_2050 = global_projections.loc[
    (global_projections["Scenario"] == "Baseline")
    & (global_projections["Year"] == 2050),
    "Projected_Emissions_Mt"
].iloc[0]

global_projections["Difference_vs_Baseline_Mt"] = (
    global_projections["Projected_Emissions_Mt"]
    - global_projections[
        "Year"
    ].map(
        global_projections[
            global_projections["Scenario"] == "Baseline"
        ].set_index("Year")[
            "Projected_Emissions_Mt"
        ]
    )
)

global_projections["Reduction_vs_Baseline_%"] = (
    (
        global_projections["Projected_Emissions_Mt"]
        / global_projections["Year"].map(
            global_projections[
                global_projections["Scenario"] == "Baseline"
            ].set_index("Year")[
                "Projected_Emissions_Mt"
            ]
        )
    ) - 1
) * 100

print("=" * 70)
print("2050 SCENARIO IMPACT")
print("=" * 70)

impact_2050 = (
    global_projections[
        global_projections["Year"] == 2050
    ][
        [
            "Scenario",
            "Projected_Emissions_Mt",
            "Difference_vs_Baseline_Mt",
            "Reduction_vs_Baseline_%"
        ]
    ]
    .sort_values("Scenario")
)

display(
    impact_2050.style.format({
        "Projected_Emissions_Mt": "{:,.2f}",
        "Difference_vs_Baseline_Mt": "{:,.2f}",
        "Reduction_vs_Baseline_%": "{:.2f}%"
    })
)
```
